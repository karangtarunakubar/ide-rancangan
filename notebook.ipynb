# Summary & Analysis: Struktur Organisasi Karang Taruna

Notebook ini memuat data struktur organisasi, ringkasan program kerja, indikator kinerja, serta analisis statistik menggunakan **Pandas**.


## 1. Import Library & Loading Data


```python
import pandas as pd
import io

# Membaca data CSV dari file local
df = pd.read_csv('struktur_organisasi.csv')

# Menampilkan 10 baris pertama
print("--- 10 Baris Pertama Data ---")
df.head(10)
```

## 2. Ringkasan Statistik Dataset


```python
# Menampilkan informasi struktur dataframe dan metrik utama
print("--- Informasi Dataset ---")
print(f"Total Baris Data: {len(df)}")
print(f"Total Kolom: {len(df.columns)}")

summary_stats = pd.DataFrame({
    'Metrik': [
        'Total Unit/Section',
        'Jumlah Sub-Section',
        'Jumlah Role/Jabatan Unik',
        'Jumlah Program Kerja Terdefinisi',
        'Jumlah Indikator Terdefinisi'
    ],
    'Nilai': [
        df['Section'].nunique(),
        df['Subsection'].nunique(),
        df['Role'].nunique(),
        len(df[df['Programs'] != '-']),
        len(df[df['Indicators'] != '-'])
    ]
})

display(summary_stats)
```

## 3. Ringkasan Per Section / Bidang Organisasi


```python
# Pengelompokan data berdasarkan Section
section_summary = df.groupby('Section').agg(
    Jumlah_Jabatan=('Role', 'count'),
    Jumlah_Program=('Programs', lambda x: sum(x != '-')),
    Jabatan_Terkait=('Role', lambda x: ", ".join(x.unique()))
).reset_index()

display(section_summary)
```

## 4. Matriks Program Kerja & Indikator Kinerja 14 Bidang (B1 - B14)


```python
# Menampilkan ringkasan khusus bidang B1 hingga B14
bidang_overview = df[df['Subsection'] == 'Overview'][['Section', 'Role Description', 'Programs', 'Indicators', 'Notes']]

print(f"Total Bidang Operasional: {len(bidang_overview)}")
display(bidang_overview)
```

## 5. Distribusi Peran (Roles) Dalam Organisasi


```python
# Distribusi frekuensi untuk setiap Role
role_distribution = df['Role'].value_counts().reset_index()
role_distribution.columns = ['Nama Role', 'Jumlah']

display(role_distribution)
```

## 6. Ekspor Ringkasan ke Excel


```python
# Menyimpan hasil analisis ke file Excel dengan beberapa sheet
output_excel = 'ringkasan_struktur_organisasi.xlsx'

with pd.ExcelWriter(output_excel, engine='openpyxl') as writer:
    df.to_excel(writer, sheet_name='Data Mentah', index=False)
    summary_stats.to_excel(writer, sheet_name='Statistik Utama', index=False)
    section_summary.to_excel(writer, sheet_name='Ringkasan Section', index=False)
    bidang_overview.to_excel(writer, sheet_name='Overview Bidang B1-B14', index=False)

print(f"File ringkasan berhasil disimpan sebagai '{output_excel}'")
```
